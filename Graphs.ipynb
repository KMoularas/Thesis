```python
dfBDI = pd.read_excel("Code Datasets.xlsx", sheet_name="BDI")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(dfBDI['Date'], dfBDI['Price'], color='blue', linewidth=1)
plt.title('Monthly Baltic Dry Index (BDI)')
plt.xlabel('Date')
plt.ylabel('BDI')
plt.savefig("BDIchart.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
dfVREBITDA = pd.read_excel("Code Datasets.xlsx", sheet_name="Voyage Revenues & EBITDA")
```

```python
vr = dfVREBITDA['Voyage Revenues']
ebidta = dfVREBITDA['EBITDA']
x = np.arange(len(dfVREBITDA['Year']))
width = 0.35

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, vr, width, label='Voyage Revenues')
plt.bar(x + width/2, ebidta, width, label='EBITDA')
plt.xticks(x, dfVREBITDA['Year'])
plt.title('Voyage Revenues and EBITDA per Year')
plt.xlabel('Year')
plt.ylabel('Voyage Revenues & EBITDA (USD thousands)')
plt.legend()

plt.savefig("VREBITDAchart.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
dfBDITCE = pd.read_excel("Code Datasets.xlsx", sheet_name="BDI & TCE 2021-2025")

x = np.arange(len(dfBDITCE['Year']))

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(x, dfBDITCE['BDI AVG'], color='blue', label='BDI', linewidth=1)
ax1.set_xlabel("Quarter")
ax1.set_ylabel("BDI")
ax1.tick_params(axis="y", labelcolor='blue')

ax1.set_xticks(x)
ax1.set_xticklabels(dfBDITCE["Quarter"], rotation=45, ha="right")

ax2= ax1.twinx()
ax2.plot(x, dfBDITCE['TCE ($)'],  color='red', label='TCE ($/day)', linewidth=1)
ax2.set_ylabel("TCE ($/day)")
ax2.tick_params(axis="y",labelcolor='red')

plt.title("BDI vs TCE Rates")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right")

fig.tight_layout()

plt.savefig("bdi_tce_chart.png", dpi=300, bbox_inches="tight")

plt.show()
```
